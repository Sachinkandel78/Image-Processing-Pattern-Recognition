# Lab 6: Image Segmentation: Contour Detection

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load image
image_path = "shiva.jpg"     
image = cv2.imread(image_path)

# Convert to RGB for display
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
#load image
img = cv2.imread("shiva.jpg", cv2.IMREAD_GRAYSCALE)

#display function
def display(title,img,cmap='gray'):
    plt.figure(figsize=(8,8))
    plt.title(title)
    plt.imshow(img,cmap=cmap)
    plt.axis()

display("Gray Image",img)

segmented_image = cv2.Canny(img,100,256)
display("Segmented Image",segmented_image)
```
