# Lab 7: Digital Image Restoration,Noise Reduction and filtering 

```python
# Lab 7: Digital Image Restoration, Noise Reduction and Filtering

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load Image
img = cv2.imread("Pokhara.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Apply Filters
mean = cv2.blur(img, (5,5))
median = cv2.medianBlur(img, 5)
gaussian = cv2.GaussianBlur(img, (5,5), 0)

# Add Salt & Pepper Noise
noisy = img.copy()

num_noise = 1000

# White pixels
for i in range(num_noise):
    x = np.random.randint(0, img.shape[0])
    y = np.random.randint(0, img.shape[1])
    noisy[x, y] = [255, 255, 255]

# Black pixels
for i in range(num_noise):
    x = np.random.randint(0, img.shape[0])
    y = np.random.randint(0, img.shape[1])
    noisy[x, y] = [0, 0, 0]

# Restore Image
restored = cv2.medianBlur(noisy, 5)

# Display Images Side by Side
plt.figure(figsize=(18,10))

plt.subplot(2,3,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(mean)
plt.title("Mean Filter")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(noisy)
plt.title("Noisy Image")
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(restored)
plt.title("Restored Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
