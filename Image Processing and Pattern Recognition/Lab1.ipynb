```python
import cv2
import matplotlib.pyplot as plt
```

```python
#load RGB image
image_path = "image.jpeg"
bgr_image = cv2.imread(image_path)
```

```python
rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(rgb_image)
plt.axis('off')  # hides the axis numbers
plt.show()
```

```python
plt.imshow(bgr_image)
plt.axis('off')
plt.show()
```

```python
grey_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2GRAY)

# Display
plt.imshow(grey_image, cmap="gray")
```

```python
grey_image = cv2.cvtColor(rgb_image, cv2.COLOR_RGB2GRAY)

# Display
plt.imshow(grey_image, cmap="grey")
plt.axis('off')  # hides the axis numbers
plt.show()
```

```python


# Display side by side
fig, axes = plt.subplots(1, 2, figsize=(10, 10))

axes[0].imshow(rgb_image)
axes[0].set_title("Original Image")


axes[1].imshow(grey_image, cmap='gray')  # cmap='gray' is important!
axes[1].set_title("Grayscale Image")    
plt.tight_layout()

plt.show()
```
