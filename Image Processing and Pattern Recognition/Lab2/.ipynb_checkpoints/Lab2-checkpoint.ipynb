```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ── Functions to display images using matplotlib ──


def show_result(original, transformed, title):
    """Display original and transformed image side by side"""
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    plt.imshow(original, cmap='gray')
    plt.title("Original Grayscale")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(transformed, cmap='gray')
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
#load RGB image
image_path = "bus.jpeg"
bgr_image = cv2.imread(image_path)
```

```python
bgr_image = cv2.imread("bus.jpeg")

# Check if loaded
if bgr_image is None:
    print("Image not found!")
else:
    rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)
    plt.imshow(rgb_image)
    plt.show()
```

```python
gray_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2GRAY)

# Display
plt.imshow(gray_image, cmap="gray")
```

```python
# ============================================================
# CELL 3: Point Operation
# ============================================================
# A point operation transforms each pixel independently.
# Every pixel in the image is processed using the same formula.
# Here we increase brightness by adding a constant value (50)
# to every pixel: s = r + 50
# cv2.add() safely clips values at 255 to avoid overflow/wrapping.

point_op = cv2.add(gray_image, 50)

# Display using show_result function
show_result(gray_image, point_op, "Point Operation (+50 Brightness)")
```

```python
# Digital Negative — invert pixel values: s = L-1 - r
# ============================================================
# CELL 4: Digital Negative
# ============================================================
# Inverts every pixel value to produce the photographic negative.
# Dark regions become bright and bright regions become dark.
# Formula: s = 255 - r  (for every pixel r in the image)
# Useful for enhancing white/gray details embedded in dark regions.

digital_negative = 255 - gray_image

# Display using show_result function
show_result(gray_image, digital_negative, "Digital Negative")
```

```python
# ============================================================
# CELL 5: Intensity Slicing
# ============================================================
# Highlights only the pixels whose intensity falls within
# a specific range [low=100, high=200].
# Pixels inside the range → white (255)
# Pixels outside the range → black (0)
# Useful for isolating objects within a known brightness range,
# e.g., highlighting bone structures in X-ray images.

# Create a blank black image of the same size as gray_image
intensity_sliced = np.zeros_like(gray_image)

# Set pixels to white only if they fall within [100, 200]
intensity_sliced[(gray_image >= 100) & (gray_image <= 200)] = 255

# Display using show_result function
show_result(gray_image, intensity_sliced, "Intensity Slicing [100–200]")
```

```python
# ============================================================
# CELL 6: Thresholding
# ============================================================
# Converts the grayscale image into a binary black-and-white image.
# A threshold value T=127 (midpoint of 0–255) is used.
# Pixels above T  → white (255)
# Pixels below T  → black (0)
# Formula: s = 255 if r > T, else s = 0
# Commonly used in image segmentation to separate
# objects from the background.

_, thresholded = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)

# Display using show_result function
show_result(gray_image, thresholded, "Thresholding (T=127)")
```

```python
# ============================================================
# CELL 7: Contrast Stretching
# ============================================================
# Improves contrast of a low-contrast image by stretching
# the pixel intensity range to cover the full spectrum [0, 255].
# If the image only uses a narrow range (e.g., 80–180),
# this remaps those values to span the complete 0–255 range.
# Formula: s = (r - r_min) / (r_max - r_min) * 255
# Makes subtle details more visible by maximizing tonal range.

r_min = gray_image.min()   # darkest pixel value in the image
r_max = gray_image.max()   # brightest pixel value in the image

# Normalize and scale pixel values to [0, 255]
contrast_stretched = ((gray_image - r_min) / (r_max - r_min) * 255).astype(np.uint8)

# Display using show_result function
show_result(gray_image, contrast_stretched, "Contrast Stretching")
```

```python
# ============================================================
# CELL 8: Summary — All Transformations in One Figure
# ============================================================
# Recomputes all transformations and displays them together
# in a single 2x3 grid for easy side-by-side comparison.
# Gives a complete visual overview of all graylevel transformations.

# Recompute all transformations
point_op         = cv2.add(gray_image, 50)
digital_negative = 255 - gray_image
intensity_sliced = np.zeros_like(gray_image)
intensity_sliced[(gray_image >= 100) & (gray_image <= 200)] = 255
_, thresholded   = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
r_min, r_max     = gray_image.min(), gray_image.max()
contrast_stretched = ((gray_image - r_min) / (r_max - r_min) * 255).astype(np.uint8)

# Store all results with their titles
results = {
    "Original"            : gray_image,
    "Point Op (+50)"      : point_op,
    "Digital Negative"    : digital_negative,
    "Intensity Slicing"   : intensity_sliced,
    "Thresholding (T=127)": thresholded,
    "Contrast Stretching" : contrast_stretched,
}

# Plot all results in a 2-row x 3-column grid
plt.figure(figsize=(18, 8))

for i, (title, img) in enumerate(results.items()):
    plt.subplot(2, 3, i + 1)
    plt.imshow(img, cmap='gray')
    plt.title(title, fontsize=12)
    plt.axis('off')

# Overall title for the summary figure
plt.suptitle("Lab 2: Graylevel Transformations — Summary",
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
