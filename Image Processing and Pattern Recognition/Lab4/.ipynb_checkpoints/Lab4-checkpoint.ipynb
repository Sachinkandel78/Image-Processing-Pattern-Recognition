Lab4: Visualising the effects of diiferent smoothing filters on image (mean,median,Guassian)


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#load RGB image
image_path = "Pokhara.jpg"
bgr_image = cv2.imread(image_path)
```

```python
rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(rgb_image)
plt.axis('off')  # hides the axis numbers
plt.show()
```

```python
mean = cv2.blur(rgb_image, (5,5))

plt.imshow(mean)
plt.title("Mean Filter")
plt.axis("off")
plt.show()
```

```python
median = cv2.medianBlur(rgb_image,5)

plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
gaussian = cv2.GaussianBlur(rgb_image, (5,5), 0)

plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
fig, ax = plt.subplots(2,2, figsize=(10,8))

ax[0,0].imshow(rgb_image)
ax[0,0].set_title("Original")
ax[0,0].axis("off")

ax[0,1].imshow(mean)
ax[0,1].set_title("Mean Filter")
ax[0,1].axis("off")

ax[1,0].imshow(median)
ax[1,0].set_title("Median Filter")
ax[1,0].axis("off")

ax[1,1].imshow(gaussian)
ax[1,1].set_title("Gaussian Filter")
ax[1,1].axis("off")

plt.tight_layout()
plt.show()
```
